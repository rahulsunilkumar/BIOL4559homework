## Preferential Attachment

```python
import numpy as np


def select_proportional_to_degree(degrees, m):
    names = np.array(list(degrees))
    d = np.array(list(degrees.values()), dtype=float)
    if sum(d) == 0:
        print(
            f"There are no existing nodes so the following are all connected to one another: {names[:m]}"
        )
    else:
        p = d / d.sum()
        print(
            f"The new node is connected to the following existing nodes: {np.random.choice(names, size=m, replace=False, p = p).tolist()}"
        )
```

```python
degrees = {
    "person1": 0,
    "person2": 0,
    "person3": 0,
    "person4": 0,
    "person5": 0,
    "person6": 0,
    "person7": 0,
    "person8": 0,
}

select_proportional_to_degree(degrees, 2)
```

## Stochastic Block Model

```python
names = [
    "person1",
    "person2",
    "person3",
    "person4",
    "person5",
    "person6",
    "person7",
    "person8",
]

# possible edges
for i, j in combinations(names, 2):
    print(i, j)
```
